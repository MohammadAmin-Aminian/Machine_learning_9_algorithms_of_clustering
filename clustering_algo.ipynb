# Nine clustering algorithms — version 2

Run in repository root. All models share standardized inputs; warnings remain visible.

```python
import pandas as pd
import matplotlib.pyplot as plt
from clustering import prepare_data, models, summarize

raw = pd.read_csv("marketing_campaign.csv", sep="\t")
SAMPLE_SIZE = 400
if SAMPLE_SIZE is not None:
    raw = raw.sample(n=min(SAMPLE_SIZE, len(raw)), random_state=42)
data, features = prepare_data(raw)
data.head()
```

## KMeans

Partitions observations by squared distance to centroids.

```python
model = models()["KMeans"]
labels = model.fit_predict(features)
print(summarize(features, labels))
plt.figure(figsize=(7, 4))
plt.scatter(data["Age"], data["Income"], c=labels, cmap="tab20", s=14)
plt.xlabel("Age (reference year 2023)")
plt.ylabel("Income")
plt.title("KMeans")
plt.show()
```
